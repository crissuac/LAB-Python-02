### ***1. Definir las siguientes funciones usando lambdas.***

# **Ejercicio 6 - Minimax**

#### Juego NIM utilizando el algoritmo Minimax

1. Descripción del problema
El juego consiste en una pila de 10 piedras. En cada turno, el jugador puede retirar 1, 2 o 3 piedras.

MAX (Computadora): juega utilizando el algoritmo Minimax.

MIN (Humano): juega ingresando su movimiento mediante input().

Condición de victoria: pierde el jugador que se encuentre con 0 piedras y no pueda realizar un movimiento.

Primer turno: comienza MIN (humano).

#### 2. Código

```python

# ==========================================
# JUEGO NIM UTILIZANDO EL ALGORITMO MINIMAX
# MAX = Computadora
# MIN = Humano
# ==========================================

# Función Minimax
def minimax(piedras, turno_max):

    # Caso base: no quedan piedras
    if piedras == 0:
        if turno_max:
            return -1  # Pierde MAX
        else:
            return 1   # Gana MAX

    # Turno de MAX (Computadora)
    if turno_max:
        mejor_valor = -float("inf")

        for movimiento in range(1, min(3, piedras) + 1):
            valor = minimax(piedras - movimiento, False)
            mejor_valor = max(mejor_valor, valor)

        return mejor_valor

    # Turno de MIN (Humano)
    else:
        mejor_valor = float("inf")

        for movimiento in range(1, min(3, piedras) + 1):
            valor = minimax(piedras - movimiento, True)
            mejor_valor = min(mejor_valor, valor)

        return mejor_valor


# Función para determinar el mejor movimiento de la CPU
def mejor_movimiento(piedras):

    mejor_valor = -float("inf")
    mejor_jugada = 1

    for movimiento in range(1, min(3, piedras) + 1):
        valor = minimax(piedras - movimiento, False)

        if valor > mejor_valor:
            mejor_valor = valor
            mejor_jugada = movimiento

    return mejor_jugada


# ==========================================
# SIMULACIÓN DEL JUEGO
# ==========================================

def jugar_nim():

    piedras = 10

    print("===================================")
    print("       JUEGO NIM - MINIMAX")
    print("===================================")
    print("Piedras iniciales:", piedras)
    print("Puedes retirar 1, 2 o 3 piedras.")
    print("El jugador que no pueda mover pierde.")
    print("El primer turno es del HUMANO.")
    print("===================================")

    # MIN comienza
    turno_humano = True

    while piedras > 0:

        print("\nPiedras restantes:", piedras)
        print("O" * piedras)

        # TURNO DEL HUMANO (MIN)
        if turno_humano:

            print("\n--- TURNO DEL HUMANO (MIN) ---")

            while True:
                try:
                    movimiento = int(input("¿Cuántas piedras deseas retirar (1-3)?: "))

                    if movimiento >= 1 and movimiento <= 3 and movimiento <= piedras:
                        break
                    else:
                        print("Movimiento inválido. Intenta nuevamente.")

                except ValueError:
                    print("Debes ingresar un número entero.")

            piedras -= movimiento

            print("El humano retiró", movimiento, "piedra(s).")

            # Si no quedan piedras, pierde el humano
            if piedras == 0:
                print("\n¡La COMPUTADORA (MAX) GANA!")
                break

            turno_humano = False

        # TURNO DE LA COMPUTADORA (MAX)
        else:

            print("\n--- TURNO DE LA COMPUTADORA (MAX) ---")

            movimiento = mejor_movimiento(piedras)

            print("La computadora retira", movimiento, "piedra(s).")

            piedras -= movimiento

            # Si no quedan piedras, pierde la computadora
            if piedras == 0:
                print("\n¡EL HUMANO (MIN) GANA!")
                break

            turno_humano = True

    print("\n===================================")
    print("          FIN DEL JUEGO")
    print("===================================")


# Ejecutar el juego
jugar_nim()
```

#### 3. Explicación del algoritmo Minimax

El algoritmo Minimax permite que la computadora analice los posibles movimientos y seleccione el que le proporcione el mejor resultado.

En este caso:

MAX (CPU): busca maximizar el resultado, intentando ganar.

MIN (Humano): busca minimizar el resultado, intentando que la computadora pierda.

Caso base: cuando las piedras llegan a cero, el jugador que realizó el último movimiento pierde.

Evaluación: se asigna 1 cuando gana MAX y -1 cuando pierde MAX.

Decisión: MAX selecciona el movimiento con el valor más alto y MIN selecciona el movimiento con el valor más bajo.

El humano interactúa mediante input(), mientras que la computadora toma sus decisiones automáticamente. El juego comienza con el humano y termina cuando se retiran todas las piedras.

Nota importante: con 10 piedras y jugando correctamente, la computadora puede garantizar su victoria si el humano comete un error. La estrategia consiste en dejar al oponente con 1, 5 o 9 piedras al finalizar el turno de la CPU.
